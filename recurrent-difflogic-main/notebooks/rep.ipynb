```python
# Scale JSON parameters with a log-based factor in a Jupyter Notebook

import json
import numpy as np

# Define parameter ranges for scaling (min, max)
PARAM_RANGES = {
    # "max_vocab_size": (500, 32000),
    "embedding_dim": (4, 1536),
    "n_layers_sizes": (16, 8000),
    "l_layers_sizes": (16, 8000),
    "k_layers_sizes": (128, 72000),
    "p_layers_sizes": (256, 128000),
    "m_layers_sizes": (256, 320000),
    "group_factor": (1, 32),
}

def log_interpolate(factor, min_val, max_val):
    """
    Interpolate between min_val and max_val in log space according to factor in [0, 1].
    factor=0 => min_val, factor=1 => max_val
    """
    min_log = np.log(min_val)
    max_log = np.log(max_val)
    val_log = min_log + (max_log - min_log) * factor
    return int(np.exp(val_log))

def scale_params(json_dict, factor):
    """
    Scale the parameters in json_dict according to factor in [0, 1].
    Modifies the json_dict in place.
    """
    # Tokenizer params
    tk_params = json_dict.get("tokenizer", {}).get("params", {})

    
    # Model params
    model_params = json_dict.get("model", {}).get("params", {})
    # Scalar values
    for param in ["embedding_dim", "group_factor"]:
        if param in model_params:
            model_params[param] = log_interpolate(factor, *PARAM_RANGES[param])
    # List-valued layer sizes
    for param in ["n_layers_sizes", "l_layers_sizes", "k_layers_sizes", "p_layers_sizes", "m_layers_sizes"]:
        if param in model_params:
            min_val, max_val = PARAM_RANGES[param]
            sizes = model_params[param]
            scaled_size = log_interpolate(factor, min_val, max_val)
            model_params[param] = [scaled_size for _ in sizes]
    return json_dict

# ---- Example usage ----
for i in range(11):

    # Load JSON file
    input_json_path = "/itet-stor/sbuehrer/net_scratch/recurrent-difflogic/configs/wmt/unsynced_recurrent_difflogic.json"
    with open(input_json_path, "r") as f:
        config = json.load(f)

    # Choose a factor between 0 and 1
    factor = i/10

    # Scale the parameters
    scaled_config = scale_params(config, factor)

    # Save scaled JSON
    output_json_path = f"/itet-stor/sbuehrer/net_scratch/recurrent-difflogic/configs/wmt/unsynced_recurrent_difflogicC_{factor:.2f}.json"
    with open(output_json_path, "w") as f:
        json.dump(scaled_config, f, indent=2)

    # Print the scaled values for inspection
    print("Scaled parameters:")
    for param, (min_val, max_val) in PARAM_RANGES.items():
        if param in scaled_config.get("tokenizer", {}).get("params", {}):
            print(f"{param}: {scaled_config['tokenizer']['params'][param]}")
        elif param in scaled_config.get("model", {}).get("params", {}):
            print(f"{param}: {scaled_config['model']['params'][param]}")
        elif param in scaled_config.get("model", {}).get("params", {}):
            print(f"{param}: {scaled_config['model']['params'][param]}")
        else:
            print(f"{param}: not found in config")
```
